```python
import plotly.graph_objects as go
import networkx as nx
import json
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt
import networkx as nx
```

```python
# static graph
with open("dag.json", "r") as f:
    _dag = json.load(f)
dag = {int(k):v for k,v in _dag.items()}

g = nx.Graph(dag)

nx.draw(g, with_labels=True)
plt.show()
```
