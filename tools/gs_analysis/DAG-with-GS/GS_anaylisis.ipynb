```python
# global GS
with open("gs.json", "r") as f:
    gs_cache = json.load(f)
gs = {int(k):v for k,v in gs_cache.items()}

x = sorted(gs.keys())
y = [gs[k] for k in x]
plt.plot(x,y)

plt.xlabel("Round")
plt.ylabel("GS")
plt.title("GS of virtual global model based on all tips in each round")
```
